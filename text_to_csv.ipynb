The target condition values are cycle-wise annotated in �profile.txt� (tab-delimited). As before, the row number represents the cycle number. The columns are</br>

1: Cooler condition / %:
- 3: close to total failure
- 20: reduced effifiency
- 100: full efficiency

2: Valve condition / %:
- 100: optimal switching behavior
- 90: small lag
- 80: severe lag
- 73: close to total failure

3: Internal pump leakage:
- 0: no leakage
- 1: weak leakage
- 2: severe leakage

4: Hydraulic accumulator / bar:
- 130: optimal pressure
- 115: slightly reduced pressure
- 100: severely reduced pressure
- 90: close to total failure

5: stable flag:
- 0: conditions were stable
- 1: static conditions might not have been reached yet


Each text file contains row representing cycles and columns representing datapoints on each cycles. For simplicity we are taking mean of datapoints of each cycle to represent the dataset more clearly.

![load_d](function_flow.png)

We have created a function name **load_data** which accepts **columns** parameter. This parameter contains an array of file names of particular sensor data.<br>
This function iterates over each text files given in **columns** parameter and converts it into pandas dataframe.

```python
def load_data(columns):
    """
    The function converts data from text files into a pandas DataFrame. Each column in the DataFrame corresponds to a text file, and the values are the mean of the data in those files.
    :param columns: A list of column names corresponding to the text files to be processed.
    :return: Pandas DataFrame containing the processed data.
    """
    data = [np.genfromtxt(os.path.join('archive', f'{col}.txt')).mean(axis=1) 
            for col in columns]
    return pd.DataFrame(data, index=columns).T
```

```python
pressure_sensor_data = ['PS1', 'PS2', 'PS3', 'PS4', 'PS5', 'PS6'] # Pressure Sensor Data
pressure_df = load_data(pressure_sensor_data)

temperature_sensor_data = ['TS1', 'TS2', 'TS3', 'TS4'] # Temperature Sensor Data
temperature_df = load_data(temperature_sensor_data)

volume_sesnor_data = ['FS1', "FS2"] # Volume Sensor Data
volume_df = load_data(volume_sesnor_data)

power_sensor_data = ['EPS1'] # Power Sensor Data
power_df = load_data(power_sensor_data)

vibration_sensor_data = ['VS1'] # Vibration Sensor Data
vibration_df = load_data(vibration_sensor_data)

cooling_efficiency_data = ['CE'] # Cooling Efficiency Data
cooling_efficiency_df = load_data(cooling_efficiency_data)

cooling_power_data = ['CP'] # Cooling Power Data
cooling_power_df = load_data(cooling_power_data)

efficiency_factor = ['SE'] # Efficiency Factor Data
efficiency_factor_df = load_data(efficiency_factor)

# Target condition values
profile_df = pd.DataFrame(np.genfromtxt(os.path.join('archive', 'profile.txt')), columns=['Cooler_Condition', 'Valve_Condition', 'Internal_Pump_Leakage', 'Hydraulic_Accumulator', 'Stable_Flag'])
```

Now we have all sensors data and condition data stored in their respective variables in the form of pandas dataframe.<br>
Next we will merge this stored data into one single datafram using pandas **pd.concat** function.

```python
# Combine all DataFrames into a single DataFrame
final_df = pd.concat([pressure_df, temperature_df, volume_df, power_df, vibration_df, cooling_efficiency_df, cooling_power_df, efficiency_factor_df, profile_df], axis=1)

# Printing first five rows of our dataframe
final_df.head()
```

Alright, now we have our final dataframe to work on. Let's go and save it in form of CSV.

```python
final_df.to_csv("Final.csv", index=False)
```
